```python
import numpy as np;
import pandas as pd;
from sklearn.model_selection import train_test_split;
import warnings;
warnings.filterwarnings("ignore");

# to compare our model's accuracy with sklearn model 
from sklearn.linear_model import LogisticRegression 

class LogisticRegressionInternal:
    def __init__(self, learningRate, epochs):
        self.learningRate = learningRate;
        self.iterations = epochs;
        self.Weights = None;
        self.Bias = None;

    def fit(self, X, Y):
        n, f = X.shape; #no_of_training_samples, features
        self.Weights = np.zeros(f);
        self.Bias = 0; #can works as int, or vector of zeros;

        for i in range(self.iterations):
            cost = self.update_weights(X, Y);
            #if (i % 100 == 0):
               # print('Cost: {}'.format(cost))

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z));

    #Y_Hat or Hypothetical function / theta(X)
    def predict(self, X):
        Z = np.dot(X, self.Weights) + self.Bias;
        Y_Hat = np.where(self.sigmoid(Z) > 0.5, 1, 0);
        return Y_Hat;

    def update_weights(self, X, Y):
        n = X.shape[0];
        Y_Hat = self.predict(X);
        diff = Y_Hat - Y;
        dW = np.dot(X.T, diff) / n;
        db = np.sum(diff) / n;

        cost = -(np.dot(Y, np.log(Y_Hat)) - np.dot(1 - Y, np.log(1 - Y_Hat))).mean();

        self.Weights = self.Weights - self.learningRate * dW;
        self.Bias = self.Bias - self.learningRate * db;

        return cost;

df = pd.read_csv("../datasets/diabetes.csv");
X = np.array(df[["Pregnancies", "Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI", "DiabetesPedigreeFunction", "Age"]]);
Y = np.array(df["Outcome"]);

# Splitting dataset into train and test set

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 1 / 3, random_state = 0);

model = LogisticRegressionInternal(0.001, 10000);
model.fit(X_train, Y_train);
print("#  Trained features values #");

print(model.Weights);
print(model.Bias);

#Testing Model
sklearnModel = LogisticRegression();
sklearnModel.fit( X_train, Y_train)
print("sklearn features", sklearnModel.coef_);
print("sklearn bias", sklearnModel.intercept_);

# Prediction on test set 
Y_pred = model.predict(X_test); #own model

Y_sklearnPred = sklearnModel.predict(X_test);


# measure performance     
correctly_classified = 0    
correctly_classified_sklearn = 0
cnt = 0;

for i in range(np.size(Y_pred)):
    if Y_test[i] == Y_pred[i]:
        correctly_classified = correctly_classified + 1;

    if Y_test[i] == Y_sklearnPred[i]:
        correctly_classified_sklearn = correctly_classified_sklearn + 1;

    cnt += 1;

print( "Accuracy on test set by our model       :  ", ( correctly_classified / cnt ) * 100 ) 
print( "Accuracy on test set by sklearn model   :  ", (correctly_classified_sklearn / cnt ) * 100 ) 


        
```
