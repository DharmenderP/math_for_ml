```python
pe = torch.zeros(block_size, n_embd)
pos = torch.arange(0, block_size, dtype=torch.float).unsqueeze(1)
pos
```

```python
import math
```

```python
div = torch.exp(torch.arange(0, n_embd, 2).float() * (-math.log(10000.0) / n_embd))
div.shape
```

```python
div
```

```python
pe[:, 0::2] = torch.sin(pos * div)
        # Apply cosine to odd indices in the positional encoding
pe[:, 1::2] = torch.cos(pos * div)

        # Add an extra dimension for batching (batch_size, seq_len, d_model)
pe = pe.unsqueeze(0)
```
